```python
import pandas as pd

url = "https://raw.githubusercontent.com/sharmaroshan/SECOM-Detecting-Defected-Items/master/uci-secom.csv"
df = pd.read_csv(url)

print(df.shape)
```

```python
df.head()
```

```python
print(df['Pass/Fail'].value_counts())
```

```python
total = len(df)
fail  = (df['Pass/Fail'] == 1).sum()

print("전체 제품:", total)
print("불량 제품:", fail)
print("불량률:", round(fail / total * 100, 2), "%")
```

```python
print("무조건 '정상'이라고만 답했을 때 정확도:",
      round((total - fail) / total * 100, 2), "%")
```

```python
sensors = df.drop(columns=['Time', 'Pass/Fail'])

missing = sensors.isna().mean() * 100

print("센서 총 개수:", sensors.shape[1])
print("빈 값이 하나라도 있는 센서:", (missing > 0).sum())
print("절반 이상 비어 있는 센서:", (missing > 50).sum())
print("값이 항상 똑같은 센서:", (sensors.nunique() <= 1).sum())
```

```python
print(missing.sort_values(ascending=False).head(10).round(1))
```
